```python
import os
import signal
import subprocess
import threading
import time

def run_gssha(MODEL_DIR, PROJECT_FILE):
    exe_path = os.path.join(MODEL_DIR, 'gssha.exe')
    prj_path = os.path.join(MODEL_DIR, PROJECT_FILE)

    if not os.path.exists(exe_path):
        raise FileNotFoundError(f"GSSHA executable not found: {exe_path}")
    if not os.path.exists(prj_path):
        raise FileNotFoundError(f"Project file not found: {prj_path}")

    print(f"📦 Running GSSHA:\n  Executable: {exe_path}\n  Project File: {prj_path}")

    stop_requested = threading.Event()

    # ----------------------------
    # Keyboard listener
    # ----------------------------
    def keyboard_listener():
        while not stop_requested.is_set():
            cmd = input("\nType 'q' and press Enter to safely stop GSSHA: ").strip().lower()
            if cmd == "q":
                stop_requested.set()

    threading.Thread(target=keyboard_listener, daemon=True).start()

    # ----------------------------
    # Start GSSHA
    # ----------------------------
    process = subprocess.Popen(
        [exe_path, prj_path],
        cwd=MODEL_DIR,
        stdout=subprocess.PIPE,
        stderr=subprocess.PIPE,
        text=True,
        creationflags=subprocess.CREATE_NEW_PROCESS_GROUP
    )

    print("✅ GSSHA started.")

    # ----------------------------
    # Main monitoring loop
    # ----------------------------
    while process.poll() is None:

        if stop_requested.is_set():

            print("\n⚠ Requesting graceful shutdown...")

            try:
                process.send_signal(signal.CTRL_BREAK_EVENT)

                process.wait(timeout=30)

                print("✅ GSSHA exited gracefully.")

            except subprocess.TimeoutExpired:

                print("⚠ Graceful shutdown timed out.")
                print("❌ Forcing termination...")

                process.terminate()

            break

        time.sleep(0.5)

    stdout, stderr = process.communicate()

    # ----------------------------
    # Save log
    # ----------------------------
    with open(os.path.join(MODEL_DIR, "gssha_log.txt"), "w") as f:
        f.write("---- STDOUT ----\n")
        f.write(stdout or "")
        f.write("\n---- STDERR ----\n")
        f.write(stderr or "")

    if process.returncode == 0:
        print("✅ GSSHA finished successfully.")
    else:
        print(f"⚠ GSSHA exited with return code {process.returncode}.")
```

```python
def read_GSSHA_dep(folder_path, filename):
    """
    Reads a GSSHA .dep file.

    Returns
    -------
    dict
        {
            "timesteps": list,
            "depths": list of numpy arrays
        }
    """

    dep_file = os.path.join(folder_path, filename)

    timesteps = []
    depths = []

    current_depths = []
    reading = False

    with open(dep_file, "r") as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            # New timestep
            if line.startswith("TS"):

                if reading:
                    depths.append(
                        np.asarray(current_depths, dtype=float)
                    )

                current_depths = []

                timestep = float(line.split()[2])

                timesteps.append(timestep)

                reading = True

                continue

            if line == "ENDDS":
                break

            if reading:
                current_depths.append(float(line))

    # save last timestep
    if current_depths:
        depths.append(np.asarray(current_depths, dtype=float))

    return {
        "timesteps": timesteps,
        "depths": depths
    }

def calculate_GSSHA_inundated_area(
    dep_data,
    cell_size,
    depth_threshold=0.01
):
    """
    Calculates inundated area statistics for each timestep.

    Parameters
    ----------
    dep_data : dict
        Output from read_GSSHA_dep()

    cell_size : float
        Grid cell size (same units as DEM)

    depth_threshold : float, optional
        Minimum depth considered inundated.

    Returns
    -------
    pandas.DataFrame

        timestep
        inundated_cells
        inundated_area
        inundated_area_change
    """

    import numpy as np
    import pandas as pd

    timesteps = dep_data["timesteps"]
    depths = dep_data["depths"]

    cell_area = cell_size ** 2

    inundated_cells = []
    inundated_area = []

    for depth in depths:

        wet_cells = np.count_nonzero(depth > depth_threshold)

        inundated_cells.append(wet_cells)
        inundated_area.append(wet_cells * cell_area)

    inundated_area_change = [0.0]

    for i in range(1, len(inundated_area)):
        inundated_area_change.append(
            inundated_area[i] - inundated_area[i-1]
        )

    df = pd.DataFrame({
        "timestep": timesteps,
        "inundated_cells": inundated_cells,
        "inundated_area": inundated_area,
        "inundated_area_change": inundated_area_change
    })

    return df
```
